```python
from qecdec.experiments import RotatedSurfaceCode_Memory
from qecdec.experiments import StimFileExperiment
import numpy as np
from pathlib import Path
```

```python
for basis in ["Z"]:
    for d in [5, 7, 9, 11]:
        for p in [0.001]:
            expmt = RotatedSurfaceCode_Memory(
                d=d,
                rounds=d,
                basis=basis,
                data_qubit_error_rate=p,
                meas_error_rate=p,
            )
            np.save(
                f"RotatedSurfaceCode_Phenom_d={d}_rounds={d}_basis={basis}_error_rate={p}_chkmat",
                expmt.chkmat,
            )
            np.save(
                f"RotatedSurfaceCode_Phenom_d={d}_rounds={d}_basis={basis}_error_rate={p}_prior",
                expmt.prior,
            )
```

```python
for basis in ["Z"]:
    for d in [5, 7, 9, 11]:
        for p in [0.001]:
            expmt = RotatedSurfaceCode_Memory(
                d=d,
                rounds=d,
                basis=basis,
                data_qubit_error_rate=p,
                prep_error_rate=p,
                meas_error_rate=p,
                gate1_error_rate=p,
                gate2_error_rate=p,
            )
            np.save(
                f"RotatedSurfaceCode_CircuitLevel_d={d}_rounds={d}_basis={basis}_error_rate={p}_chkmat",
                expmt.chkmat,
            )
            np.save(
                f"RotatedSurfaceCode_CircuitLevel_d={d}_rounds={d}_basis={basis}_error_rate={p}_prior",
                expmt.prior,
            )
```

```python
circuit_dir = Path("../../circuits")
```

```python
for basis in ["Z"]:
    for p in [0.001]:
        circuit_file = (
            circuit_dir
            / "BB_144_12_12_CircuitLevel"
            / f"d=12_rounds=12_basis={basis}"
            / f"error_rate={p}.stim"
        )
        expmt = StimFileExperiment.load_from_file(circuit_file, detector_basis=basis)
        np.save(
            f"BB_144_12_12_CircuitLevel_d=12_rounds=12_basis={basis}_error_rate={p}_chkmat",
            expmt.chkmat,
        )
        np.save(
            f"BB_144_12_12_CircuitLevel_d=12_rounds=12_basis={basis}_error_rate={p}_prior",
            expmt.prior,
        )
```
