```python
from qecdec.experiments import RotatedSurfaceCode_Memory
import numpy as np
```

```python
for basis in ["Z"]:
    for d in [5, 7, 9, 11]:
        for p in [0.001]:
            expmt = RotatedSurfaceCode_Memory(
                d=d,
                rounds=d,
                basis=basis,
                data_qubit_error_rate=p,
                meas_error_rate=p,
            )
            np.save(
                f"RotatedSurfaceCode_Phenom_d={d}_rounds={d}_basis={basis}_error_rate={p}_chkmat",
                expmt.chkmat,
            )
            np.save(
                f"RotatedSurfaceCode_Phenom_d={d}_rounds={d}_basis={basis}_error_rate={p}_prior",
                expmt.prior,
            )
```
